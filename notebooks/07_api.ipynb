```python
import numpy as np
import pandas as pd
import joblib

df = pd.read_csv("../data/feature_matrix.csv")
model = joblib.load("../artifacts/isolation_forest.pkl")

# Keep only the columns the model was trained on, in the same order
X = df[model.feature_names_in_]

scores = -model.score_samples(X)

cutoff = np.percentile(scores, 97)
print("Score cutoff:", cutoff)
print("Share above cutoff:", (scores > cutoff).mean())   # should be about 0.03
```
